# Financial Doc Analyzer - interactive tour
Start the servers first: `make mock` and `make app`.

```python
import requests, json
APP='http://localhost:9025'
def qa(q, fid=None): return requests.post(f'{APP}/qa', json={'question': q, 'filing_id': fid}).json()
def rs(fid): return requests.post(f'{APP}/risksummary', json={'filing_id': fid}).json()
print(json.dumps(qa('What are the cybersecurity risks for ACMR-2023?'), indent=2))
```

```python
print(json.dumps(rs('HLXR_2022'), indent=2))
```
